# Vision-Diff-Physics Results Viz
Run this after `python train.py --config configs/default.yaml`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

run_dir = Path('../results/default_run')
metrics = pd.read_csv(run_dir / 'metrics.csv')
metrics.tail()
```

```python
plt.figure(figsize=(7, 4))
plt.plot(metrics['epoch'], metrics['train_traj_mse'], label='train')
plt.plot(metrics['epoch'], metrics['val_traj_mse'], label='val')
plt.yscale('log')
plt.xlabel('epoch')
plt.ylabel('trajectory MSE')
plt.legend()
plt.show()
```
